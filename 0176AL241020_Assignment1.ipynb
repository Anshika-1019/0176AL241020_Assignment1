```python
#CONDITIONAL STATEMENT

# Q1. Electricity Bill Calculator

units = int(input("Enter electricity units: "))

if units <= 100:
    bill = units * 5

elif units <= 300:
    bill = (100 * 5) + (units - 100) * 7

else:
    bill = (100 * 5) + (200 * 7) + (units - 300) * 10

if bill > 2000:
    bill = bill + (bill * 5 / 100)

print("Total Electricity Bill =", bill)
```

```python

# Q3. LCM and HCF Using Loops

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

hcf = 1

for i in range(1, a + 1):
    if a % i == 0 and b % i == 0:
        hcf = i

lcm = a if a > b else b

while lcm % a != 0 or lcm % b != 0:
    lcm = lcm + 1

print("HCF =", hcf)
print("LCM =", lcm)
```

```python

# Q5. String Compression

text = input("Enter a string: ")

result = ""
count = 1

for i in range(1, len(text)):
    if text[i] == text[i - 1]:
        count += 1
    else:
        result += text[i - 1] + str(count)
        count = 1

result += text[-1] + str(count)

print("Compressed String:", result)
```

```python
# FUNCTION AND RECURSION

# Q1. Factorial Using Function
def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact = fact * i
    return fact

num = int(input("Enter a number: "))
print("Factorial:", factorial(num))
```

```python
# Q2. Sum of First N Natural Numbers
def natural_sum(n):
    if n == 0:
        return 0
    return n + natural_sum(n - 1)

num = int(input("Enter a number: "))
print("Sum:", natural_sum(num))
```

```python
# Q3. Reverse a String Using Recursion
def reverse_string(s):
    if len(s) == 0:
        return s
    return reverse_string(s[1:]) + s[0]

text = input("Enter a string: ")
print("Reversed String:", reverse_string(text))
```

```python
# Q4. Power of a Number Using Recursion
def power(x, n):
    if n == 0:
        return 1
    return x * power(x, n - 1)

x = int(input("Enter base: "))
n = int(input("Enter power: "))

print("Result:", power(x, n))
```

```python
# OBJECT ORIENTED PROGRAMMING

# Q1. Multilevel Inheritance

class Vehicle:
    def start(self):
        print("Vehicle starts")

class Car(Vehicle):
    def drive(self):
        print("Car is driving")

class ElectricCar(Car):
    def charge(self):
        print("Electric car is charging")

e = ElectricCar()
e.start()
e.drive()
e.charge()
```

```python
# Q2. Multiple Inheritance

class Teacher:
    def teach(self):
        print("Teaching students")

class Researcher:
    def research(self):
        print("Doing research")

class Professor(Teacher, Researcher):
    def work(self):
        print("Working as a professor")

p = Professor()
p.teach()
p.research()
p.work()
```

```python
# Q3. Hierarchical Inheritance

class Shape:
    def display(self):
        print("This is a shape")

class Circle(Shape):
    def area(self, r):
        print("Circle area:", 3.14 * r * r)

class Rectangle(Shape):
    def area(self, l, b):
        print("Rectangle area:", l * b)

class Triangle(Shape):
    def area(self, b, h):
        print("Triangle area:", 0.5 * b * h)

c = Circle()
r = Rectangle()
t = Triangle()

c.display()
c.area(5)

r.display()
r.area(4, 6)

t.display()
t.area(4, 5)
```

```python
# Q4. Method Overriding

class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")

a = Animal()
d = Dog()
c = Cat()

a.sound()
d.sound()
c.sound()
```

```python
# Q5. Operator Overloading

class Complex:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def __add__(self, other):
        return Complex(self.real + other.real,
                       self.imag + other.imag)

    def display(self):
        print(self.real, "+", self.imag, "i")

c1 = Complex(3, 4)
c2 = Complex(2, 5)

c3 = c1 + c2
c3.display()
```

```python
# Q7. ATM System Using Encapsulation

class ATM:
    def __init__(self, pin, balance):
        self.__pin = pin
        self.__balance = balance

    def check_balance(self, pin):
        if pin == self.__pin:
            print("Balance:", self.__balance)
        else:
            print("Incorrect PIN")

    def withdraw(self, pin, amount):
        if pin == self.__pin:
            if amount <= self.__balance:
                self.__balance -= amount
                print("Withdrawal successful")
                print("Remaining balance:", self.__balance)
            else:
                print("Insufficient balance")
        else:
            print("Incorrect PIN")

a = ATM(1234, 5000)
a.check_balance(1234)
a.withdraw(1234, 1000)
```
